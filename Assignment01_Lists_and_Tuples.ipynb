```python
#27.  How to sort a list in ascending and descending order without using the sort function?
my_list = [13, 5, 87, 42, 79, 34]

def sort_list(my_list, ascending=True):

    for i in range(len(my_list)):
        for j in range(i + 1, len(my_list)):
            if ascending and my_list[i] > my_list[j]:
                my_list[i], my_list[j] = my_list[j], my_list[i]

            elif not ascending and my_list[i] < my_list[j]:
                my_list[i], my_list[j] = my_list[j], my_list[i]

    return my_list

print("Ascending order:", sort_list(my_list.copy(), True))
print("Descending order:", sort_list(my_list.copy(), False))
```

```python
#35. What is the difference between copy and deep copy?
#copy is also called shallow copy and other one is deep copy
#Copy: Creates a new list, but if the original list contains nested lists, both lists share those inner lists.
#Changes to an inner list in the copy will also affect the original.
#Deep copy: Creates a completely independent copy of the list, including nested lists,2D Array. Changes made to the copied list will not affect the original list.

my_list = [[1, 2], [3, 4]]
new_list = my_list.copy()
new_list[0][0] = 84
print(my_list)
print(new_list)
```

```python
#deep copy example:
import copy
my_list = [[1, 2],[3, 4]]
new_list = copy.deepcopy(my_list)

new_list[0][0] = 84
print(my_list)
print(new_list)
```

```python
#38. Given a list of integers, create a new list where each element is replaced by the next largest
#number in the original list. If there is no larger number, replace the element with -1.
#input_list = [4, 2, 6, 8, 1, 2]
#output_list = [6, 6, 8, -1, 2, -1]
input_list = [4, 2, 6, 8, 1, 2]
output_list = []

for i in range(len(input_list)):
    next_greater = -1

    for j in range(i + 1, len(input_list)):
        if input_list[j] > input_list[i]:
            next_greater = input_list[j]
            break

    output_list.append(next_greater)
print("Output list:", output_list)
```

```python
#39. Write a function to reverse the elements within each tuple in a list of tuples.
#tuples = [(1, 2), (3, 4), (5, 6)]
#Output: [(2, 1), (4, 3), (6, 5)]
def reverse_tuples(tuples):
    output_list = []
    for item in tuples:
        reversed_tuple = item[::-1]
        output_list.append(reversed_tuple)

    return output_list
tuples = [(1, 2), (3, 4), (5, 6)]
result = reverse_tuples(tuples)
print("Output:", result)
```

```python
#40. Write a function to add corresponding elements of tuples in two lists of tuples
# list1 = [(1, 2), (3, 4)]
#list2 = [(5, 6), (7, 8)]
#Output: [(6, 8), (10, 12)]
list1 = [(1, 2), (3, 4)]
list2 = [(5, 6), (7, 8)]
def add_tuples(list1, list2):
    output_list = []

    for i in range(len(list1)):
        first_sum = list1[i][0] + list2[i][0]
        second_sum = list1[i][1] + list2[i][1]

        output_list.append((first_sum, second_sum))
    return output_list
result = add_tuples(list1, list2)
print("Output:", result)
```
