```python
#13. Group a list of words into lists of anagrams using a dictionary. Print the list of lists where each inner list contains words that are anagrams of each other.
#Input: ["listen", "silent", "enlist", "hello", "lemon", "melon", "debitcard", "badcredit"]
#Anagram groups: [['listen', 'silent', 'enlist'], ['hello'], ['lemon', 'melon'], ['debitcard', 'badcredit']]

input = ["listen", "silent", "enlist", "hello", "lemon", "melon", "debitcard", "badcredit"]
anagram_groups = {}

for word in input:
    key = "".join(sorted(word))

    if key in anagram_groups:
        anagram_groups[key].append(word)
    else:
        anagram_groups[key] = [word]

result = list(anagram_groups.values())

print("Anagram groups:", result)

```

```python
#14. Merge two dictionaries. If there are common keys, sum their values. Print the merged dictionary.
dict1 = {"a": 10, "b": 20, "c": 30}
dict2 = {"b": 50, "c": 60, "d": 40}

merged_dict = dict1.copy()
for key, value in dict2.items():
    if key in merged_dict:
        merged_dict[key] += value
    else:
        merged_dict[key] = value

print("Merged dictionary:", merged_dict)
```

```python
#16. Find mirror characters in a string using a dictionary.
#For example, 'A' mirrors to 'Z', 'B' mirrors to 'Y', Input string: "Hello World"

mirror_string = {
    'A': 'Z', 'B': 'Y', 'C': 'X', 'D': 'W',
    'E': 'V', 'F': 'U', 'G': 'T', 'H': 'S',
    'I': 'R', 'J': 'Q', 'K': 'P', 'L': 'O',
    'M': 'N', 'N': 'M', 'O': 'L', 'P': 'K',
    'Q': 'J', 'R': 'I', 'S': 'H', 'T': 'G',
    'U': 'F', 'V': 'E', 'W': 'D', 'X': 'C',
    'Y': 'B', 'Z': 'A'
}
input_string = "Hello World"
result = ""

for char in input_string:
    if char.upper() in mirror_string:
        mirror_char = mirror_string[char.upper()]

        if char.islower():
            result += mirror_char.lower()
        else:
            result += mirror_char
    else:
        result += char

print("Mirror string:", result)
```

```python
#20. Count the number of vowels in a given string and store the counts in a dictionary.
input_string = "Hello World"
vowels = "aeiou"
vowel_count = {}

for char in input_string.lower():
    if char in vowels:
        if char in vowel_count:
            vowel_count[char] += 1
        else:
            vowel_count[char] = 1

print("Vowel counts:", vowel_count)
```

```python
#26. Find the minimum and maximum elements in a set without using the built-in min() and max() functions.
set1 = {45, 12, 78, 23, 9, 56}
minimum = None
maximum = None
for num in set1:
    if minimum is None or num < minimum:
        minimum = num

    if maximum is None or num > maximum:
        maximum = num

print("Minimum element:", minimum)
print("Maximum element:", maximum)
```

```python
#28. Find the intersection of two sets with and without using built-in methods.
set1 = {10, 20, 30, 40}
set2 = {30, 40, 50, 60}

# with built-in function. We can also write using '&' operator (result = s1 & s2)
result = set1.intersection(set2)
print("Intersection:", result)

#without built in
result = set()
for num in set1:
    if num in set2:
        result.add(num)
print("Intersection:", result)
```

```python
#29. Find the symmetric difference of two sets with and without using built-in methods.

set1 = {10, 20, 30, 40}
set2 = {30, 40, 50, 60}
#with
result = set1.symmetric_difference(set2)
print("Symmetric Difference:", result)

#without
result = set()
for num in set1:
    if num not in set2:
        result.add(num)
for num in set2:
    if num not in set1:
        result.add(num)

print("Symmetric Difference:", result)
```

```python
#30. Perform set operations (union, intersection, difference) on sets containing both integers and strings without using built-in methods.
#Set 1: {1, 2, 'apple', 'orange'}
#Set 2: {2, 3, 'orange', 'banana'}
set1 = {1, 2, 'apple', 'orange'}
set2 = {2, 3, 'orange', 'banana'}

# Union
union_set = set1.copy()
for item in set2:
    if item not in union_set:
        union_set.add(item)
print("Union:", union_set)

# Intersection
intersection_set = set()
for item in set1:
    if item in set2:
        intersection_set.add(item)
print("Intersection:", intersection_set)


# Difference (Set 1 - Set 2)
difference_set = set()
for item in set1:
    if item not in set2:
        difference_set.add(item)
print("Difference:", difference_set)
```
